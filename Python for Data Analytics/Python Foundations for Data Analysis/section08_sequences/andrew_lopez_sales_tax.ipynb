### Sales Tax

We need to calculate the sales taxes for our customer that made three transactions in `multi_order_customers` below.

(The customer that made three transactions was customer C00001). Run this cell to create the list.

```python
multi_order_customers = [
    [1799.94, 29.98, 99.99],
    [15.98, 119.99],
    [24.99, 24.99],
    [649.99, 99.99],
    [599.99, 399.97]
]
```

Create a tuple from the first element in the `multi_order_customers`.

Unpack this tuple into the variables `transaction1`, `transaction2`, and `transaction3`.

Once you've have separated the transactions, calculate the tax of .08 rounded to 2 decimals and print it.

```python
c00001Transactions = tuple(multi_order_customers[0])
print(f"this is C00001's transactions: {c00001Transactions}")

transaction1,transaction2, transaction3 = c00001Transactions

print(f"transaction1: ${transaction1} - transaction2: ${transaction2} - transaction3: ${transaction3}")
```

```python
print(f"The sales tax for C00001's transactions are\ntransaction1: ${round(transaction1 *.08,2)} - transaction2: ${round(transaction2 *.08,2)} - transaction3: ${round(transaction3 *.08,2)}")
```
